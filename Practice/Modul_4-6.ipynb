# Module 4: Decomposition, Abstraction, and Functions

## Brief review

A function gives a name to one computation. Parameters receive input, local variables store temporary state, and `return` supplies a result to the caller. Decomposition assigns separate responsibilities to separate functions.

```python
# Complete example: define, call, and use a returned value.
def calculate_good_units(total_units, defective_units):
    return total_units - defective_units


good_units = calculate_good_units(520, 18)
weekly_estimate = good_units * 5

assert good_units == 502
assert weekly_estimate == 2510
print(f"Good units: {good_units}")
```

```python
# Complete example: one function coordinates smaller functions.
def calculate_material_cost(quantity, unit_cost):
    return quantity * unit_cost


def calculate_labor_cost(hours, hourly_rate):
    return hours * hourly_rate


def calculate_total_cost(quantity, unit_cost, hours, hourly_rate):
    material = calculate_material_cost(quantity, unit_cost)
    labor = calculate_labor_cost(hours, hourly_rate)
    return material + labor


assert calculate_total_cost(10, 8_000, 2, 40_000) == 160_000
```

# Module 5: Tuples, Lists, and Aliasing

## Brief review

A tuple is an ordered immutable sequence. A list is an ordered mutable sequence. Assignment can make two names refer to the same list. Use a copy when the new list must change independently.

```python
# Complete example: a function returns one tuple with three results.
def summarize_values(values):
    return min(values), max(values), sum(values)


summary = summarize_values([128, 121, 135, 130])
minimum, maximum, total = summary

assert summary == (121, 135, 514)
assert minimum == 121
assert maximum == 135
assert total == 514
```

```python
# Complete example: an alias shares a list; a copy has a new outer list.
original = [120, 125, 130]
alias = original
independent = original.copy()

alias[0] = 118
independent[1] = 128

assert original == [118, 125, 130]
assert alias is original
assert independent == [120, 128, 130]
assert independent is not original
```

# Module 6: Dictionaries and Recursion

## Brief review

A dictionary maps unique keys to values. Recursion solves a problem through a smaller call to the same function. A recursive function needs a base case and progress toward that case.

```python
# Complete example: accumulate repeated inventory records.
receipts = [("BOLT", 12), ("NUT", 8), ("BOLT", 7), ("NUT", 4)]
inventory = {}

for item, quantity in receipts:
    inventory[item] = inventory.get(item, 0) + quantity

assert inventory == {"BOLT": 19, "NUT": 12}
print(inventory)
```

```python
# Complete example: base case, smaller call, and returned result.
def sum_from_one(number):
    if number == 0:
        return 0
    return number + sum_from_one(number - 1)


assert sum_from_one(0) == 0
assert sum_from_one(4) == 10
```
