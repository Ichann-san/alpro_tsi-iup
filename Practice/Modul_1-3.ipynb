# Module 2: Basic Branching and Iteration

## Brief review

An `if` statement runs code only when a condition is true. `elif` and `else` add alternative paths. A `for` loop processes items from a list, while a `while` loop repeats while its condition remains true.

```python
# Complete example: classify one result.
actual_units = 480
target_units = 500

if actual_units > target_units:
    status = "ABOVE TARGET"
elif actual_units == target_units:
    status = "ON TARGET"
else:
    status = "BELOW TARGET"

print(status)
assert status == "BELOW TARGET"
```

```python
# Complete example: counter and accumulator patterns.
daily_output = [480, 510, 495, 525]
target_units = 500
successful_days = 0
total_output = 0

for units in daily_output:
    total_output += units
    if units >= target_units:
        successful_days += 1

assert successful_days == 2
assert total_output == 2010
print(f"Successful days: {successful_days}")
```

# Module 3: Basic String Manipulation

## Brief review

Strings store text. Indexing selects one character, slicing extracts a substring, and string methods return cleaned or transformed strings. `split()` creates a list of fields, while `join()` combines strings.

```python
# Complete example: inspect and clean one string.
raw_label = "  hydraulic   pump  "
words = raw_label.split()
clean_label = " ".join(words).title()
code_label = "-".join(words).upper()

assert clean_label == "Hydraulic Pump"
assert code_label == "HYDRAULIC-PUMP"
assert raw_label == "  hydraulic   pump  "
print(code_label)
```
