# Module 2 — Branching and Iteration

## Brief review

An `if`–`elif`–`else` chain selects one mutually exclusive path. A `for` loop consumes an iterable. A `while` loop repeats while its condition remains true. Every condition-controlled loop requires a termination argument: some state must move toward a condition that ends the loop.

`break` exits the nearest loop. `continue` skips the remaining statements in the current iteration.

```python
# Complete review example: classification and finite iteration
readings = [72, -1, 85, 101, 64]
accepted = []

for reading in readings:
    if reading < 0 or reading > 100:
        continue
    if reading >= 80:
        label = "high"
    elif reading >= 60:
        label = "normal"
    else:
        label = "low"
    accepted.append(f"{reading}:{label}")

assert accepted == ["72:normal", "85:high", "64:normal"]
print("Module 2 review checks passed.")
```
