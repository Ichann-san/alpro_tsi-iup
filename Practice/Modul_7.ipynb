## 1. Testing and assertions

A test compares an actual result with an expected result. An assertion makes the comparison automatic.

Start with ordinary input, then include a boundary case and zero. A passing assertion produces no output. A failed assertion raises `AssertionError`.

```python
# Complete example: calculate expected values independently.
def calculate_cost(quantity, unit_cost):
    return quantity * unit_cost


assert calculate_cost(3, 5_000) == 15_000
assert calculate_cost(1, 5_000) == 5_000
assert calculate_cost(0, 5_000) == 0

print("Cost checks passed.")
```

## 3. Exceptions

`int("many")` raises `ValueError`. A `try` block contains the operation that may fail, and a matching `except` block defines the response.

An invalid value can also be rejected explicitly using `raise ValueError("message")`.

Use `raise` for required input rules and `assert` for test expectations. Assertions may be disabled with Python's `-O` option.

```python
# Complete example: handle invalid numeric text.
raw_value = "many"

try:
    parsed_value = int(raw_value)
    print(f"Parsed value: {parsed_value}")
except ValueError:
    print("The text must represent an integer.")
```

```python
# Complete example: reject a value that violates a rule.
def check_hours(hours):
    if hours < 0:
        raise ValueError("Hours cannot be negative.")
    return hours


assert check_hours(8) == 8

error_detected = False
try:
    check_hours(-1)
except ValueError:
    error_detected = True

assert error_detected, "Negative hours must be rejected."
```
